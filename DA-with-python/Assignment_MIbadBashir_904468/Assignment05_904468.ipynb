## Question 1: Working with Lists

```python
# Create a list of 8 integers
numbers = [12, 7, 25, 8, 30, 15, 42, 5]

# Loop 1: print every number in the list
print("All numbers:")
for num in numbers:
    print(num)

# Loop 2: add up all the numbers
total = 0
for num in numbers:
    total = total + num
print("Sum of all numbers:", total)

# Loop 3: print only the even numbers (number divisible by 2)
print("Even numbers:")
for num in numbers:
    if num % 2 == 0:
        print(num)
```

## Question 2: Dictionary Basics

```python
# Dictionary with a list of 3 marks inside it
student = {
    "name": "Ibad",
    "age": 20,
    "course": "Data Analytics with Python",
    "marks": [78, 65, 82]
}

# Print all keys and their values
print("Student details:")
for key, value in student.items():
    print(f"{key}: {value}")

# Use a loop to calculate the average of the marks
marks_total = 0
for mark in student["marks"]:
    marks_total = marks_total + mark
average = marks_total / len(student["marks"])
print("Average marks:", round(average, 2))

# Pass if average is 50 or more, otherwise Fail
if average >= 50:
    print("Result: Passed")
else:
    print("Result: Failed")
```

## Question 3: List of Dictionaries

```python
# List where each employee is a dictionary
employees = [
    {"name": "Ahmed", "department": "Sales", "salary": 55000},
    {"name": "Sara", "department": "Finance", "salary": 72000},
    {"name": "Bilal", "department": "IT", "salary": 65000},
    {"name": "Zara", "department": "HR", "salary": 48000}
]

# Print details of all employees
print("Employee details:")
for emp in employees:
    print(f"Name: {emp['name']}, Department: {emp['department']}, Salary: {emp['salary']}")

# Find the employee with the highest salary (comparison logic)
highest = employees[0]
for emp in employees:
    if emp["salary"] > highest["salary"]:
        highest = emp
print(f"Highest salary: {highest['name']} ({highest['salary']})")

# Calculate the total salary expense
total_salary = 0
for emp in employees:
    total_salary = total_salary + emp["salary"]
print("Total salary expense:", total_salary)
```

## Question 4: While Loop with User Input

```python
input = make_input(['5', '12', 'abc', '3', '9', '-1']) if DEMO else real_input


entered_numbers = []   # list to store the numbers

print("Enter numbers one by one. Type -1 to stop.")
number = 0
while number != -1:
    text = input("Enter a number: ")
    try:
        number = int(text)          # convert text to a whole number
    except ValueError:
        print("Invalid input, please enter a whole number.")
        continue
    if number != -1:                # -1 is only the stop signal, so do not store it
        entered_numbers.append(number)

# After the loop ends
if len(entered_numbers) == 0:
    print("No numbers were entered.")
else:
    largest = entered_numbers[0]
    smallest = entered_numbers[0]
    for n in entered_numbers:
        if n > largest:
            largest = n
        if n < smallest:
            smallest = n
    print("Complete list:", entered_numbers)
    print("Largest number:", largest)
    print("Smallest number:", smallest)
```

## Question 5: Word Frequency Counter

```python
input = make_input(['apple banana apple orange banana apple']) if DEMO else real_input


sentence = input("Enter a sentence: ")

# Split the sentence into a list of words (lowercase so Apple and apple match)
words = sentence.lower().split()

# Dictionary to count each word
word_count = {}
for word in words:
    if word in word_count:
        word_count[word] = word_count[word] + 1   # word seen before
    else:
        word_count[word] = 1                      # first time seeing the word

# Print the word frequency
print("Word frequency:")
for word, count in word_count.items():
    print(f"{word}: {count}")
```

## Question 6: Nested Loops - Multiplication Table

```python
# Outer loop = rows (1 to 5)
for i in range(1, 6):
    # Inner loop = columns (1 to 5)
    for j in range(1, 6):
        print(f"{i * j:>3}", end=" ")   # keep numbers lined up in columns
    print()                              # move to the next line after each row
```
