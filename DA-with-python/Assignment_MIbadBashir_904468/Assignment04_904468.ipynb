# Dictionaries Assignment No. 4

```python

DEMO = True

real_input = input

def make_input(values):
    it = iter(values)
    def fake_input(prompt=""):
        value = next(it)
        print(prompt + value)
        return value
    return fake_input
```

## Q1: Basic Dictionary Creation

```python
student = {
    "name": "Ibad",
    "age": 20,
    "course": "Data Analytics with Python",
    "city": "Karachi"
}
for key in student:
    print(f"{key}: {student[key]}")
```

## Q2: Adding & Updating Values

```python
car = {
    "brand": "Toyota",
    "model": "Corolla",
    "year": 2018
}
car["color"] = "White"   # add new key
car["year"] = 2022       # update existing key
print(car)
```

## Q3: Loop Through Dictionary

```python
person = {
    "name": "Ibad",
    "age": 25,
    "city": "Karachi"
}
for key, value in person.items():
    print(key, ":", value)
```

## Q4: Count Frequency of Numbers

```python
numbers = [1, 2, 2, 3, 3, 3, 4, 4, 4, 4]
frequency = {}
for num in numbers:
    if num in frequency:
        frequency[num] += 1
    else:
        frequency[num] = 1
print(frequency)
```

## Q5: Word Counter

```python
input = make_input(['the cat and the dog and the bird']) if DEMO else real_input


sentence = input("Enter a sentence: ")
word_count = {}
for word in sentence.lower().split():
    word = word.strip(".,!?;:\"'()")   # remove punctuation around words
    if word == "":
        continue
    if word in word_count:
        word_count[word] += 1
    else:
        word_count[word] = 1
print("Word frequency:")
for word, count in word_count.items():
    print(f"  {word}: {count}")
```

## Q6: Student Marks & Average

```python
marks = {"Math": 80, "English": 70, "Physics": 90}
total = 0
for subject in marks:
    total += marks[subject]
average = total / len(marks)
print("Total marks :", total)
print("Average     :", round(average, 2))
```

## Q7: Highest Scoring Student

```python
students = {
    "Ali": 85,
    "Sara": 92,
    "Ahmed": 78,
    "Zara": 95
}
top_name = None
top_score = -1
for name, score in students.items():
    if score > top_score:
        top_score = score
        top_name = name
print(f"Highest scoring student: {top_name} with {top_score} marks")
```

## Q8: Merge Dictionaries

```python
dict1 = {"a": 1, "b": 2}
dict2 = {"c": 3, "d": 4}
merged = dict1.copy()
merged.update(dict2)
print("Merged dictionary:", merged)
```

## Q9: Mini Project: Contact Book

```python
input = make_input(['Ali', '03001234567', 'Sara', '03211234567', 'Bilal', '03331234567', 'sara', 'Zain', '']) if DEMO else real_input


contacts = {}

print("Add 3 contacts:")
for i in range(3):
    name = input(f"Contact {i + 1} name : ").strip()
    phone = input(f"Contact {i + 1} phone: ").strip()
    contacts[name.lower()] = phone

print("\nSearch contacts (press Enter without typing to stop)")
while True:
    search = input("Search by name: ").strip().lower()
    if search == "":
        break
    if search in contacts:
        print("Phone:", contacts[search])
    else:
        print("Contact not found")
```
