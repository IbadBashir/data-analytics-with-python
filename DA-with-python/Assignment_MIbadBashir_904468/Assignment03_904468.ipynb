# List Assignment No. 3

```python

DEMO = True

real_input = input

def make_input(values):
    it = iter(values)
    def fake_input(prompt=""):
        value = next(it)
        print(prompt + value)
        return value
    return fake_input

def convert(value):
    """Convert text to int or float if possible, otherwise keep it as text."""
    try:
        return int(value)
    except ValueError:
        try:
            return float(value)
        except ValueError:
            return value


def get_list(prompt="Enter list elements separated by spaces: "):
    return [convert(item) for item in input(prompt).split()]


def get_number_list(prompt="Enter numbers separated by spaces: "):
    while True:
        try:
            return [float(x) if "." in x else int(x) for x in input(prompt).split()]
        except ValueError:
            print("Please enter numbers only.")


def get_int(prompt):
    while True:
        try:
            value = int(input(prompt))
            if value > 0:
                return value
            print("Please enter a number greater than 0.")
        except ValueError:
            print("Please enter a valid whole number.")


def get_matrix(rows, cols, name="Matrix"):
    print(f"Enter elements of {name} ({rows} x {cols}):")
    matrix = []
    for i in range(rows):
        while True:
            try:
                row = [float(x) if "." in x else int(x)
                       for x in input(f"  Row {i + 1} ({cols} numbers): ").split()]
                if len(row) != cols:
                    print(f"  Please enter exactly {cols} numbers.")
                    continue
                matrix.append(row)
                break
            except ValueError:
                print("  Please enter numbers only.")
    return matrix


def print_matrix(matrix):
    for row in matrix:
        print("  ".join(f"{x:>5}" for x in row))
```

## Question 1: Print alternate elements of a list

```python
input = make_input(['1 2 3 4 5']) if DEMO else real_input

lst = get_list()
print("Original list     :", lst)
alternate = []
for i in range(0, len(lst), 2):
    alternate.append(lst[i])
print("Alternate elements:", alternate)
```

## Question 2: Reverse a list (without reverse())

```python
input = make_input(['10 20 30 40']) if DEMO else real_input

lst = get_list()
print("Original list:", lst)
reversed_list = []
for i in range(len(lst) - 1, -1, -1):
    reversed_list.append(lst[i])
print("Reversed list:", reversed_list)
```

## Question 3: Largest number (without max())

```python
input = make_input(['4 9 2 7 15 3']) if DEMO else real_input

lst = get_number_list()
if not lst:
    print("The list is empty.")
else:
    largest = lst[0]
    for num in lst:
        if num > largest:
            largest = num
    print("List          :", lst)
    print("Largest number:", largest)
```

## Question 4: Rotate list elements (last element moves to first)

```python
input = make_input(['1 2 3 4 5']) if DEMO else real_input

lst = get_list()
print("Original list:", lst)
if len(lst) > 1:
    last = lst[-1]
    for i in range(len(lst) - 1, 0, -1):
        lst[i] = lst[i - 1]
    lst[0] = last
print("Rotated list :", lst)
```

## Question 5: Delete a given word from a string

```python
input = make_input(['stop and smell the roses', 'and']) if DEMO else real_input

text = input("Enter a string: ")
word = input("Enter the word to delete: ")
words = text.split()
result = [w for w in words if w != word]
if len(result) == len(words):
    print(f'The word "{word}" was not found in the string.')
print("New string:", " ".join(result))
```

## Question 6: Convert mm/dd/yyyy to Month dd, yyyy

```python
input = make_input(['03/12/2021']) if DEMO else real_input

months = ["January", "February", "March", "April", "May", "June",
          "July", "August", "September", "October", "November", "December"]
date = input("Enter date in mm/dd/yyyy format: ").strip()
try:
    mm, dd, yyyy = date.split("/")
    print(f"{months[int(mm) - 1]} {int(dd)}, {yyyy}")
except (ValueError, IndexError):
    print("Invalid date. Please use the mm/dd/yyyy format, e.g. 03/12/2021")
```

## Question 7: Capitalize each word (using a function)

```python
input = make_input(['stop and smell the roses.']) if DEMO else real_input

def capitalize_words(sentence):
    new_words = []
    for word in sentence.split():
        new_words.append(word[0].upper() + word[1:].lower())
    return " ".join(new_words)


sentence = input("Enter a sentence: ")
print("Result:", capitalize_words(sentence))
```

## Question 8: Sum of each row of an m x n matrix

```python
input = make_input(['3', '4', '2 11 7 12', '5 2 9 15', '8 3 10 42']) if DEMO else real_input

m = get_int("Enter number of rows (m): ")
n = get_int("Enter number of columns (n): ")
matrix = get_matrix(m, n)
print("\nMatrix:")
print_matrix(matrix)
print()
for i in range(m):
    total = 0
    for j in range(n):
        total += matrix[i][j]
    print(f"Sum of row {i + 1} = {total}")
```

## Question 9: Add two matrices of size n x m

```python
input = make_input(['2', '2', '1 2', '3 4', '5 6', '7 8']) if DEMO else real_input

n = get_int("Enter number of rows (n): ")
m = get_int("Enter number of columns (m): ")
a = get_matrix(n, m, "Matrix A")
b = get_matrix(n, m, "Matrix B")
result = []
for i in range(n):
    row = []
    for j in range(m):
        row.append(a[i][j] + b[i][j])
    result.append(row)
print("\nA + B =")
print_matrix(result)
```

## Question 10: Multiply two matrices

```python
input = make_input(['2', '2', '2', '2', '1 2', '3 4', '5 6', '7 8']) if DEMO else real_input

r1 = get_int("Enter rows of Matrix A: ")
c1 = get_int("Enter columns of Matrix A: ")
r2 = get_int("Enter rows of Matrix B: ")
c2 = get_int("Enter columns of Matrix B: ")

if c1 != r2:
    print("Multiplication not possible: columns of A must equal rows of B.")
else:
    a = get_matrix(r1, c1, "Matrix A")
    b = get_matrix(r2, c2, "Matrix B")
    result = [[0 for _ in range(c2)] for _ in range(r1)]
    for i in range(r1):
        for j in range(c2):
            for k in range(c1):
                result[i][j] += a[i][k] * b[k][j]
    print("\nA x B =")
    print_matrix(result)
```
